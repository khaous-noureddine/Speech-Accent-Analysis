```python
import pandas as pd

df = pd.read_csv("/home/nkhaous/myLLF/speech_accent/representations_distances/data/metadata_balanced_10.csv")

# 1️⃣ Identifier pays anglais avec >20 speakers
english_natives = df[df["native_language"].str.lower() == "english"]
country_counts = english_natives["country"].value_counts()
valid_countries = country_counts[country_counts > 20].index.tolist()

print("Keeping English natives from:")
print(valid_countries)

# 2️⃣ Filtrer natifs anglais
df_filtered = df.copy()
mask_english = df_filtered["native_language"].str.lower() == "english"

df_filtered = df_filtered[
    ~mask_english |
    (df_filtered["country"].isin(valid_countries))
]

# 3️⃣ Renommer english → english_country
df_filtered.loc[
    df_filtered["native_language"].str.lower() == "english",
    "native_language"
] = (
    "english_" +
    df_filtered.loc[
        df_filtered["native_language"].str.lower() == "english",
        "country"
    ].str.lower().str.replace(" ", "_")
)

# 4️⃣ Recalculer count_native_language
new_counts = df_filtered["native_language"].value_counts()
df_filtered["count_native_language"] = df_filtered["native_language"].map(new_counts)

# 5️⃣ Save
output_path = "/home/nkhaous/myLLF/speech_accent/representations_distances/data/metadata_filtered_english_split.csv"
df_filtered.to_csv(output_path, index=False)

print(f"\nSaved to: {output_path}")
print("\nUpdated counts:")
print(df_filtered["native_language"].value_counts())
```
